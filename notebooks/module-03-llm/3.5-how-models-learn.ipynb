````python
import numpy as np

def cross_entropy_loss(probabilities, target_index):
    """Computes cross-entropy loss for the target token."""
    target_prob = probabilities[target_index]
    # Add a tiny epsilon to prevent log(0)
    eps = 1e-12
    return -np.log(target_prob + eps)

# Simulated probabilities over 4 vocabulary tokens: ["the", "dog", "cat", "runs"]
vocab = ["the", "dog", "cat", "runs"]
target_idx = 2  # Target token is "cat"

# Model A is confident and correct
probs_a = np.array([0.05, 0.05, 0.85, 0.05])
loss_a = cross_entropy_loss(probs_a, target_idx)

# Model B is guessing randomly
probs_b = np.array([0.25, 0.25, 0.25, 0.25])
loss_b = cross_entropy_loss(probs_b, target_idx)

# Model C is confident but WRONG
probs_c = np.array([0.80, 0.15, 0.02, 0.03])
loss_c = cross_entropy_loss(probs_c, target_idx)

print(f"Model A (confident & right): Loss = {loss_a:.4f}")
print(f"Model B (random guess):     Loss = {loss_b:.4f}")
print(f"Model C (confidently wrong): Loss = {loss_c:.4f}")
````

````python
import numpy as np

# We want our model to predict y = 6.0 when given x = 2.0 (ideal weight = 3.0)
x = 2.0
y_true = 6.0

# Start with a random initial weight
weight = 0.5
learning_rate = 0.05

print(f"Initial weight: {weight:.4f}")

for step in range(1, 6):
    # Forward pass: prediction
    y_pred = weight * x
    
    # Mean squared error loss: (y_pred - y_true)^2
    loss = (y_pred - y_true) ** 2
    
    # Calculus gradient: d(Loss)/d(weight) = 2 * (y_pred - y_true) * x
    gradient = 2 * (y_pred - y_true) * x
    
    # Gradient descent update step
    weight = weight - (learning_rate * gradient)
    
    print(f"Step {step}: Pred={y_pred:.2f}, Loss={loss:.4f}, New Weight={weight:.4f}")
````
